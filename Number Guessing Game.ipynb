<b> Number Guessing Game using Random Library </b>

```python
import random

secret_number = random.randint(1, 100)
guess_count = 0

while True:
    guess_count += 1
    print(f"\nAttempt : {guess_count}")

    guess = int(input("Enter your guess: "))

    if guess > secret_number:
        print("Too High!")
    elif guess < secret_number:
        print("Too Low!")
    else:
        print("Congratulations! You guessed it!")
        break
        
print(f"\nYou guessed the secret number in {guess_count} moves.")
```
